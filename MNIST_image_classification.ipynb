```python
# Preprocess Data

x_train = df.drop('label', axis=1).values
y_train = df['label'].values
x_test = df_test.drop('label', axis=1).values
y_test = df_test['label'].values
```

```python
x_train
```

```python
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0
```

```python
# x_train_img = x_train.reshape(x_train.shape[0], 28, 28, 1)
# x_test_img = x_test.reshape(x_test.shape[0], 28, 28, 1)
```

```python
x_train_img = x_train.reshape(-1, 28, 28)
x_test_img = x_test.reshape(-1, 28, 28)
```

-1: Yeh Python ko batata hai ki 'number of samples' ko automatic calculate kar lo.

28, 28: Yeh phir se har image ki height aur width hai, jo ki 28 by 28 pixels hai.

```python
x_train_img
```

```python
x_train_img.shape
```

```python
y_train_cat = to_categorical(y_train, num_classes=10)
y_test_cat = to_categorical(y_test, num_classes=10)
```

# **Perceptron**

```python
perceptron = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(10, activation='softmax')
])
```

```python
perceptron.compile(optimizer='sgd', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
history_perceptron = perceptron.fit(x_train_img, y_train_cat, epochs=10, batch_size=32,
                                    validation_data=(x_test_img, y_test_cat), verbose=2)
```

```python
accuracy_perceptron = perceptron.evaluate(x_test_img, y_test_cat)[1]
print(f"Accuracy of Perceptron: {accuracy_perceptron * 100:.2f}%")
```

# **ANN**

```python
ann = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation='relu'),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])
```

```python
ann.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
history_ann = ann.fit(x_train_img, y_train_cat, epochs=10, batch_size=32,
                      validation_data=(x_test_img, y_test_cat), verbose=1)
```

```python
accuracy_ann = ann.evaluate(x_test_img, y_test_cat)[1]
print(f"Accuracy of ANN: {accuracy_ann * 100:.2f}%")
```

# **CNN**

```python
x_train_cnn = x_train.reshape(-1, 28, 28, 1)
x_test_cnn = x_test.reshape(-1, 28, 28, 1)
```

```python
cnn = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),

    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),

    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(10, activation='softmax')
])
```

```python
cnn.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
history_cnn = cnn.fit(x_train_cnn, y_train_cat, epochs=5, batch_size=32,
                     validation_data=(x_test_cnn, y_test_cat), verbose=1)
```

```python
accuracy_cnn = cnn.evaluate(x_test_cnn, y_test_cat)[1]
print(f"Accuracy of CNN: {accuracy_cnn * 100:.2f}%")
```
